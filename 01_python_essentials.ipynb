```python
def calculate_discount(price, discount_pct=10):
    """Calculate discounted price.
    Args:
        price: Original price in CZK
        discount_pct: Discount percentage (default 10%)
        Returns:
            Discounted price rounded to whole CZK
        """
    discounted = price * (1 - discount_pct/100)
    return round(discounted)
# Use it
print(calculate_discount(500)) # 450 (default 10%)
print(calculate_discount(500, 20)) # 400 (20% discount)
print(calculate_discount(500, discount_pct=15)) # 425 (named argument)
    
```

```python
def summarise_books(books):
    """Print a summary of a book collection."""
    count = len(books)
    total = sum(b["price"] for b in books)
    avg = total / count if count > 0 else 0
    cheapest = min(books, key=lambda b: b["price"])
    priciest = max(books, key=lambda b: b["price"])
    print(f" Book Collection Summary:")
    print(f" Total books: {count}")
    print(f" Total value: {total} CZK")
    print(f" Average price: {avg:.0f} CZK")
    print(f" Cheapest: {cheapest['title']} ({cheapest['price']} CZK)")
    print(f" Priciest: {priciest['title']} ({priciest['price']} CZK)")

    # Test it
    books = [
        {"title": "The Castle", "price": 299}, {"title": "Solaris", "price": 325}, {"title": "1984", "price": 189}, {"title": "Dune", "price": 520},]
        
summarise_books(books)
```

```python
def categorise_price(price):
    """Categorise a book price in to a tier."""
    if price > 400:
        return "Premium"
    elif price >= 200:
        return "Standard"
    else:
        return "Budget"
books = [{ "title": "The Castle", "price": 299}, {"title": "Solaris", "price": 325}, {"title": "1984", "price": 189}, {"title": "Dune", "price": 520},]

# Add tier to each book
for book in books:
    book["tier"] = categorise_price(book["price"])

# Check the result
for book in books:
    print(f"{book['title']:<15} {book['price']:>5} CZK -> {book['tier']}")

# Output:
# The Castle 299 CZK -> Standard
# Solaris 325 CZK -> Standard
# 1984 189 CZK -> Budget
# Dune 520 CZK -> Premium
    
```

```python
import csv
from datetime import datetime
# --- Load books
books = []
with open("data/raw/books.csv", "r", encoding="utf") as f:
    for row in csv.DictReader(f):
        row["price"] = float(row["price"]) # Convert string to number
        row["in_stock"] = row["in_stock"] == "t" #convert to boolean
        books.append(row)
print(f"Loaded {len(books)} books\n")

# --- Basic stats
prices = [b["price"] for b in books]
in_stock = [b for b in books if b["in_stock"]]
out_of_stock = [b for b in books if not b["in_stock"]]

print(" Knihy a Kafe - Book Inventory Report")
print("=" * 45)
print(f"Total books:    {len(books)}")
print(f"In stock:       {len(in_stock)}")
print(f"Out of stock:   {len(out_of_stock)}")
print(f"Price range:    {min(prices):.0f} - {max(prices):.0f} CZK")
print(f"Average price:  {sum(prices) / len(prices):.0f} CZK")

# --- Books by price tier
def categorise(price):
    if price > 400: return "Premium"
    elif price >= 200: return "Standard" 
    else: return "Budget"
tier_counts = {}
for b in books:
    tier = categorise(b["price"])
    tier_counts[tier] = tier_counts.get(tier, 0) + 1
print(f"\n Books by price tier:")
for tier, count in sorted(tier_counts.items()):
    print(f"  {tier:<10} {count} books")

# --- Top 5 most expensive
print(f"\n Top 5 Most Expensive Books:")
top5 = sorted(books, key=lambda b: b["price"], reverse=True)[:5]
for i, b in enumerate(top5, 1):
    print(f"  {i}. {b['title']:<50} {b['price']:>7.0f} CZK")

```
